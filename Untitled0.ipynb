```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

sns.set_style('whitegrid')

df = pd.read_csv('NCRB_Table_1A.1.csv')
df.head()
```

```python
features = ['Mid-Year Projected Population (in Lakhs) (2022)',
            'Rate of Cognizable Crimes (IPC) (2022)',
            'Chargesheeting Rate (2022)']

X = df[features]

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
inertia = []
k_range = range(1, 8)
for k in k_range:
    km = KMeans(n_clusters=k, random_state=42, n_init=10)
    km.fit(X_scaled)
    inertia.append(km.inertia_)

plt.figure(figsize=(7,5))
plt.plot(k_range, inertia, marker='o')
plt.xlabel('Number of Clusters (k)')
plt.ylabel('Inertia')
plt.title('Elbow Method for Optimal k')
plt.show()
```

```python
for k in range(2, 7):
    km = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = km.fit_predict(X_scaled)
    score = silhouette_score(X_scaled, labels)
    print(f'k={k}, silhouette score={score:.3f}')
```

```python
k = 3  # <-- replace with your chosen number
kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
df['Cluster'] = kmeans.fit_predict(X_scaled)
df[['State/UT', 'Cluster']].sort_values('Cluster')
```

```python
plt.figure(figsize=(8,6))
sns.scatterplot(data=df, x='Rate of Cognizable Crimes (IPC) (2022)',
                 y='Chargesheeting Rate (2022)', hue='Cluster', palette='Set2', s=100)
for _, row in df.iterrows():
    plt.annotate(row['State/UT'], (row['Rate of Cognizable Crimes (IPC) (2022)'],
                 row['Chargesheeting Rate (2022)']), fontsize=7)
plt.title('State Clusters: Crime Rate vs Chargesheeting Rate')
plt.tight_layout()
plt.show()
```
